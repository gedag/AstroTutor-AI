# Domain Corpus v1.0 — keşifsel analiz

Pipeline çıktısını (`data/final/`) yükler, dağılımlara ve rastgele örneklere bakar. Raporlardaki sayılar `scripts/08_statistics.py` ile üretilir; bu defter yalnızca gözle kontrol içindir.

```python
import pandas as pd
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
df = pd.concat([pd.read_json(ROOT / f'data/final/{s}.jsonl', lines=True) for s in ('train', 'validation', 'test')])
df.shape
```

```python
df.groupby('source').agg(docs=('document_id', 'count'), tokens=('token_count', 'sum'), median_tokens=('token_count', 'median'))
```

```python
pd.crosstab(df.subtopic, df.source)
```

```python
pd.crosstab(df.language, df.split, normalize='columns').round(3)
```

```python
# Rastgele 3 doküman: temizliğin gözle kontrolü
for _, r in df.sample(3, random_state=42).iterrows():
    print('=' * 100); print(r.document_id, r.source, r.subtopic, r.url); print(r.text[:1500])
```

```python
# En kısa ve en uzun dokümanlar
df.sort_values('token_count')[['document_id', 'source', 'title', 'token_count']].iloc[list(range(5)) + list(range(-5, 0))]
```

```python
ins_path = ROOT / 'data/final/instruction_v1.0.jsonl'
if ins_path.exists():
    ins = pd.read_json(ins_path, lines=True)
    display(pd.crosstab(ins.task_type, ins.split, margins=True))
    display(pd.crosstab(ins.task_type, ins.level))
```
